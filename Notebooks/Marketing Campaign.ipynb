```python
# 3. Drop columns not needed for the analysis
df = df.drop(columns=["Company", "Duration", "Language"])
print(df.shape)
```

```python
# 4. Clean Acquisition_Cost: strip "$" and "," then convert string -> integer
df["Acquisition_Cost"] = (
    df["Acquisition_Cost"].str.replace(r"[$,]", "", regex=True).astype(float).astype(int)
)
```

```python
df.head()
```

```python
#5. Rename ROI to a clearer label
df = df.rename(columns={"ROI": "ROI %"})
```

```python
#6. Change conversion column into percentage format 
df["Conversion_Rate"] = (df["Conversion_Rate"] * 100).round(2)
```

```python
#7.Rename Conversion_Rate to a clearer label
df = df.rename(columns={"Conversion_Rate": "Conversion %"})
```

```python
#8. Parse Date from text (day/month/year) into a real datetime
df["Date"] = pd.to_datetime(df["Date"], format="%d/%m/%Y")
```

```python
#9. Save the cleaned file
df.to_csv("marketing_campaign_dataset-cleaned.csv", index=False)
```
